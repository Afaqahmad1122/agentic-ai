```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")
response = llm.invoke("how many moons does a jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assistant, and you will answer only in 1 line"],
    ["human", "How many moons does Jupiter have?"]
])
print(response.text)
```
